Download queries(questions)

```python
from datasets import load_dataset
sub_dataset_name = "vidore_v3_computer_science"
dataset_name = f"vidore/{sub_dataset_name}"
base_path = "/upb/users/a/amirk/scratch/UPB_phd/RAG/"

subsets = ["queries"]

```

```python
for subset in subsets:
    print(f"Downloading {subset}...")
    ds = load_dataset(dataset_name, subset)
    ds.save_to_disk(f"{base_path}/{sub_dataset_name}/{subset}")
    print(f"Saved {subset} to {base_path}/{sub_dataset_name}/{subset}")
```

Download PDFs

```python
from huggingface_hub import snapshot_download

snapshot_download(
    repo_id="vidore/vidore_v3_computer_science",
    repo_type="dataset",
    local_dir=f"{base_path}/{sub_dataset_name}/pdfs",
    allow_patterns="pdfs/*"
)
```

Get first n questions

```python
from datasets import load_from_disk

ds = load_from_disk(f"{base_path}/{sub_dataset_name}/queries")

# look at structure first
print(ds)
print(ds["test"][0])  # first example

# first n questions
n = 10
questions = ds["test"][:n]["query"]
answers = ds["test"][:n]["answer"]

print("\n\n****\nQUESTIONS")
print(questions)


print("\n\n****\ANSWERS")
print(answers)
```
